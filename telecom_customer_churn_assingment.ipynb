```python
num_cols=df.select_dtypes(include=np.number).columns
num_cols
```

```python
cat_cols=df.select_dtypes(include=np.object_).columns
cat_cols
```

```python
for col in num_cols:
  if col !=target:
    df[col]=df[col].fillna(df[col].median())
```

```python
for col in cat_cols:
  df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Task:-7 dublicate records
df.duplicated().sum()
```

```python
# Task:-8 datatypes
df.dtypes
```

```python
df.head(2)
```

```python
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce")
```

```python
df["monthly_charges"] = df["monthly_charges"].fillna(
    df["monthly_charges"].median()
)

df["total_charges"] = df["total_charges"].fillna(
    df["total_charges"].median()
)
```

```python
# task:- 9 outlier
for col in num_cols:
  plt.figure(figsize=(5,6))
  sns.boxplot(data=df,y=df[col])
  plt.title(f"Boxplot of{col}")
```

```python
df['churn']
```

```python
Q1 = df["total_charges"].quantile(0.25)
Q3 = df["total_charges"].quantile(0.75)
Q1, Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_limit = Q1- 1.5*IQR
upper_limit = Q3 + 1.5*IQR
lower_limit , upper_limit
```

```python
df=df[(df["total_charges"] >= lower_limit) & (df["total_charges"] <= upper_limit) ]
```

```python
df.shape
```

```python
print(
    "Shape after cleaning:",
    df.shape
)

print(
    "Missing values:",
    int(df.isna().sum().sum())
)

print(
    "Duplicate rows:",
    int(df.duplicated().sum())
)

print(
    "Cleaning verification complete.")
```
